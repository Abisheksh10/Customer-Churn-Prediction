```python
churn_counts=df["Churn"].value_counts()
print("Count of churned and Non churned Customers")
print(churn_counts)
churn_rate=(df["Churn"]=="Yes").mean()*100
print(f"Churn rate is {churn_rate}")
```

```python
df["TotalCharges"]=pd.to_numeric(df["TotalCharges"],errors="coerce")
print(df["TotalCharges"].isna().sum())
df["TotalCharges"]=df["TotalCharges"].fillna(df["MonthlyCharges"])
```

```python
print("minimum of tenure: ",df["tenure"].min())
print("maximum of tenure: ",df["tenure"].max())
print("Mean of tenure: ",df["tenure"].mean())
print("Minimum of monthly charges: ",df["MonthlyCharges"].min())
print("Maximum of monthly charges: ",df["MonthlyCharges"].max())
print("Mean of monthly charges: ",df["MonthlyCharges"].mean())
print("Minimum of total charges: ",df["TotalCharges"].min())
print("Maximum of total charges: ",df["TotalCharges"].max())
print("Mean of total charges: ",df["TotalCharges"].mean())
```
